```python
df["Transaction_Date"] = pd.to_datetime(df["Transaction_Date"],errors="coerce")
```

```python
numeric_cols = df.select_dtypes(include=["number"]).columns
```

```python
df[numeric_cols] = df[numeric_cols].fillna( df[numeric_cols].mean())
```

```python
text_cols = df.select_dtypes(include=["object"]).columns
```

```python
df[text_cols] = df[text_cols].fillna("Unknown")
```

```python
df["Transaction_Date"] = df["Transaction_Date"].fillna(pd.Timestamp("2024-01-01"))
```

```python
df["Year"] = df["Transaction_Date"].dt.year
```

```python
df["Month"] = df["Transaction_Date"].dt.month_name()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
revenue = df.groupby("Fiscal_Year")["Revenue"].sum()
```

```python
plt.figure(figsize=(8,5))
plt.plot(revenue.index, revenue.values, marker="o")
plt.title("Revenue by Fiscal Year")
plt.xlabel("Fiscal Year")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()
```

```python
region = df.groupby("Region")["Revenue"].sum()
```

```python
plt.figure(figsize=(8,5))
plt.bar(region.index, region.values)
plt.title("Revenue by Region")
plt.xlabel("Region")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
status = df["Invoice_Status"].value_counts()
```

```python
plt.figure(figsize=(6,6))
plt.pie(status, labels=status.index, autopct="%1.1f%%")
plt.title("Invoice Status")
plt.show()
```

```python
sns.histplot(df["Revenue"], bins=40)
plt.title("Revenue Distribution")
plt.show()
```

```python
sns.boxplot(x="Region", y="Profit", data=df)
plt.xticks(rotation=45)
plt.show()
```
